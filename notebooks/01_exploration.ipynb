```python
# Q1 — Créer la SparkSession
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("TradeCorp ETL")
    .getOrCreate()
)
print(spark.version)
```

```python
# Q2 — Lire les 8 CSV
path_raw = '../data/raw/'
df_categories = spark.read.csv(f"{path_raw}categories.csv",header=True,inferSchema=True)
df_categories.show()
df_customers = spark.read.csv(f"{path_raw}customers.csv",header=True,inferSchema=True)
df_customers.show()

df_employees =  spark.read.csv(f"{path_raw}employees.csv",header=True,inferSchema=True)
df_employees.show()

df_order_details = spark.read.csv(f"{path_raw}order_details.csv",header=True,inferSchema=True)
df_order_details.show()

df_orders = spark.read.csv(f"{path_raw}orders.csv",header=True,inferSchema=True)
df_orders.show()

df_products = spark.read.csv(f"{path_raw}products.csv",header=True,inferSchema=True)
df_products.show()

df_shippers = spark.read.csv(f"{path_raw}shippers.csv",header=True,inferSchema=True)
df_shippers.show()

df_suppliers = spark.read.csv(f"{path_raw}suppliers.csv",header=True,inferSchema=True)
df_suppliers.show()
```

```python
# Q7 — Lazy evaluation
print("""
    Q7 — Lazy evaluation
    
    La lazy evaluation (évaluation paresseuse) signifie que Spark n'exécute pas immédiatement les opérations demandées, elles sont exécutées à travers des actions.
    
    Une transformation crée un nouveau DataFrame à partir d'un DataFrame existant.
    
    Exemples de transformations :
    
    filter()
    select()
    withColumn()
    
    Une action déclenche l'exécution des opérations précédentes et produit un résultat ou récupère les données.
    
    Exemples d'actions :
    
    show()
    count()
    collect()
    
    En résumé : les transformations définissent les opérations à faire, et les actions déclenchent leur exécution.
""")
```

```python
# Q7 — Lazy evaluation
print("""
Q7 — Lazy evaluation

La lazy evaluation (évaluation paresseuse) signifie que Spark n'exécute pas immédiatement les opérations demandées, elles sont exécutées à travers des actions.

Une transformation crée un nouveau DataFrame à partir d'un DataFrame.

Exemples de transformations :
filter()
select()
withColumn()

Une action déclenche l'exécution des opérations précédentes et produit un résultat ou récupère les données comme par exemple :
show()
count()
collect()

En résumé : les transformations définissent les opérations et les actions permettent de les exécuter.
""")
```

```python
# Q8 — Spark UI
print("""
Q8 — Spark UI

La Spark UI permet de suivre l’exécution des applications Spark. Dans notre cas, on observe 50 jobs terminés.

Un Job correspond à une opération déclenchée par une action Spark, par exemple count(), show() ou csv(). La Spark UI permet de voir sa durée, les stages exécutés et le nombre de tasks réalisées.

Un Stage est une étape du traitement Spark. Un job peut être composé d’un ou plusieurs stages. Les stages permettent à Spark de découper le traitement en différentes étapes d’exécution.

Une Task est une unité de travail exécutée par Spark à l’intérieur d’un stage. Les données sont réparties en partitions et chaque task traite une partition.

Dans notre Spark UI, on observe notamment que les jobs affichent 1/1 stage réussi et 1/1 task réussie. Cela signifie que, pour ces opérations, Spark a exécuté un seul stage contenant une seule task, avec une exécution réussie.

La Spark UI permet donc de visualiser et de suivre les performances et le déroulement des traitements Spark : Jobs → Stages → Tasks.
""")
```

```python
# Q9 — Spark vs Pandas
import pandas as pd
import time
start =time.time()
df_categories_pandas = pd.read_csv(f"{path_raw}categories.csv")
end = time.time()
temps_pandas = end - start
df_categories_pandas.count()
print(f"temps de lecture avec pandas {temps_pandas:.4f} en secondes")
```

```python
#spark time
start = time.time()
df_categories.count()
end =time.time()
temps_spark = end - start
print(f"temps de lecture avec pandas {temps_spark:.4f} en secondes")
```

```python
print("===== Comparaison =====")
print(f"Pandas : {temps_pandas:.4f} secondes")
print(f"Spark  : {temps_spark:.4f} secondes")

if temps_pandas < temps_spark:
    print("Pandas est plus rapide sur ce fichier.")
else:
    print("Spark est plus rapide sur ce fichier.")
```
